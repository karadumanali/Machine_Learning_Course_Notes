```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df=pd.read_csv("cyber_attack_data.csv")
```

```python
df.head()
```

```python
# src_packet_rate -> Source-side packet transmission rate
# dst_packet_rate -> Destination-side packet reception rate
# avg_payload_size -> Average size of payload in packets
# connection_duration -> Duration of the connection (in seconds)
# tcp_flag_count -> Number of TCP flag occurrences
# avg_interarrival_time -> Time between packet arrivals
# failed_login_attempts -> Number of failed login attempts
# unusual_port_activity_score -> Score representing unusual port usage
# session_entropy -> Entropy of session behavior (for anomaly detection)
# avg_response_delay -> Average delay in server response (in ms)
# attack_type -> 0 = Normal, 1 = DDoS, 2 = Port Scan
```

```python
df.info()
```

```python
X=df.drop("attack_type", axis=1)
y=df["attack_type"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=15)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(X_train, y_train)
y_pred=model.predict(X_test)
```

```python
y_pred #multi-class calismiiiiis
```

```python
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
```

```python
score=accuracy_score(y_pred, y_test)
print("score: ",score,"\n")

print(classification_report(y_pred, y_test),"\n")

print("confussion matrix: \n", confusion_matrix(y_pred, y_test))

#y_test, y_pred  ile y_pred, y_test arasinda siralama farki oluyor sonuclarda
```

# One vs Rest

```python
from sklearn.multiclass import OneVsOneClassifier, OneVsRestClassifier
#diyerek bunlari import edebilirz
```

```python
onevsone_model=OneVsOneClassifier(LogisticRegression()) #diyerek icine bir estimator verebiliriz
onevsrest_model=OneVsRestClassifier(LogisticRegression()) #ayni sekilde buna da
```

```python
onevsone_model.fit(X_train, y_train)
y_one_pred = onevsone_model.predict(X_test)

score_one=accuracy_score(y_one_pred, y_test)
print("score: ",score_one,"\n")

print(classification_report(y_one_pred, y_test),"\n")

print("confussion matrix: \n", confusion_matrix(y_one_pred, y_test))
```

```python
onevsrest_model.fit(X_train, y_train)
y_rest_pred = onevsrest_model.predict(X_test)

score_rest=accuracy_score(y_rest_pred, y_test)
print("score: ",score_rest,"\n")

print(classification_report(y_rest_pred, y_test),"\n")

print("confussion matrix: \n", confusion_matrix(y_rest_pred, y_test))
```
