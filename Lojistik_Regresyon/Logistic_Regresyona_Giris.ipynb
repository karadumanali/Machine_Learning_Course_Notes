```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv("bank_customers.csv")
```

```python
df.head()
```

#### bu data seti hazir bir data seti. Scale edilmis ,non degerler olmayan calismaya hazir ama gerceklige uygundur

### burdaki hedef column "subscribed". Yani bir kampanya var hangi musteriler buna katilim saglicak mi saglamicak mi bunu ogrenmek isteyebiliriz:

```python
df.info()
```

```python
X=df.drop("subscribed", axis=1)
y=df["subscribed"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test= train_test_split(X, y, test_size= 0.3, random_state=15)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic =LogisticRegression() #normalde verebilecegimiz cok parametre vardir ama suanlik vermicez
logistic.fit(X_train, y_train)
y_pred=logistic.predict(X_test)
```

```python
y_pred #gordugumuz gibi predict'ler 0 ve 1'lerden olusuyor
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

#bu recall, f1, precissions skoru gibi degerler geliyor "classification_report"un icerisinde
```

```python
score = accuracy_score(y_pred, y_test)
print("score: ", score,"\n")

print(classification_report(y_test, y_pred))

print("confusion_matrix: \n",confusion_matrix(y_test, y_pred))
```

# Hyper Parameter Tuning

### LogisticRegression() sift+tab yaparsak icerisinde bircok parametre cikacaktir.<br>-*-__penalty__: "l2", "l1", "elasticnet" veya "none" seceneklerinden biri kullanilabilir. l2->ridge, l1->lasso degerlerini hatirlarsin. default olarak "l2" kullanilir. Ucunu de denicez biz-*-<br><br>-*-__C__: regularizatin strenght yani ridge, lasso'daki __lambda(alpha)__ degeridir. Default 1'dir. biz 100,10,1,0.1,0.01 degerlerini denicez. En uygun senaryo hangi degerde cikacak ona bakacagiz-*-<br><br>-*-__solver__:  optimizasyon algoritmasini belirlemek icin kullanilir. loss-function'un min degerini bulmak icin matematiksel bir yontem uygular. Solver parametresi, modelin en iyi performansi elde etmesi icin parametreleri (agirliklari) nasil bulacagini belirleyen optimizasyon algoritmasidir.  solver bir modelin __ne__ yapacağını değil, __nasıl__ yapacağını belirleyen mekanizmadır. Biz butun seceneklerini kullanacagiz. Asagida mevcuttur

[Scikit-Learn/Logistic_Regression Dokumantasyonuna gitmek icin tiklayin](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LogisticRegression.html)

```python
model = LogisticRegression() 
```

```python
penaltilar = ["l1", "l2", "elasticnet"]
c_values=[100, 10, 1, 0.1, 0.01]
solvers = ["newton-cg", "libliner", "lbfgs", "sag", "saga", "newton-cholesky"]

#bunlar en onemli parametrelerdir. yukarida istenirse gidilecek link mevcuttur
```

### biz bunlari grid_search kutuphanesini kullanmak icin dict tipine donusturmemiz gerekiyor cunku o kutuphane oyle istiyor

```python
parametreler = dict(penalty=penaltilar, C=c_values, solver=solvers)
```

```python
parametreler
```

# Grid Search CV

### bir model için en iyi hiperparametreleri bulmamıza yardımcı olur. <br> adından da anlaşıldığı gibi, belirlediğiniz hiperparametre değerlerinin bir "ızgarasını" (grid) oluşturur ve bu ızgaradaki her kombinasyonu deneyerek en iyi kombinasyonu bulmaya çalışır. <br><br>Grid Search CV, manuel olarak hiperparametreleri tek tek denemekten çok daha verimlidir. Elle deneme yapmak hem zaman alır hem de en iyi kombinasyonu bulma garantisi vermez. Grid Search, sistematik bir şekilde tüm olası kombinasyonları deneyerek en iyi sonucu bulmanızı sağlar ve bu süreç otomatiktir.

## StratifiedKFold

### StratifiedKFold'un temel amacı, veri setini bölerken her bir alt kümede (fold) sınıf dağılımını korumaktır. Bu, özellikle veri setinizdeki sınıfların dengesiz olduğu durumlarda hayati önem taşır. <br><br>StratifiedKFold, veri setini k adet alt kümeye bölerken, her bir alt kümenin orijinal veri setindeki sınıf oranlarını yansıtmasını sağlar.<br><br>mesela Elinizde 1000 örnekten oluşan bir veri seti var.  Sınıf A: 900 örnek (%90).  Sınıf B: 100 örnek (%10) <br><br>__KFold (Standart Bölme):__ Veriyi rastgele 5 parçaya bölerseniz, bir veya iki parçada hiç Sınıf B örneği olmayabilir. Bu, modelin eğitimini ve testini olumsuz etkiler.<br><br>__StratifiedKFold (Katmanlı Bölme):__ Veriyi 5 parçaya bölerken, her bir parçada yaklaşık 180 Sınıf A ve 20 Sınıf B örneği olmasını sağlar. Yani, her parçanın içindeki sınıf oranı (%90 A, %10 B) korunur.

[Scikit-Learn/GridSearchCV Dokumantasyonuna gitmek icin tiklayiniz](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GridSearchCV.html)

```python
from sklearn.model_selection import GridSearchCV, StratifiedKFold
```

```python
cv =StratifiedKFold() #icerisindeki cv default=5'dir
```

### GridSearchCV() icerisine "__estimator__--> verilecek model", "__param_grid__--> deneyecegi parametreler", "__cv__--> koyulacaak cv", "__scoring__--> bu en iyi optimizasyonu __neye__ gore bulacak"

```python
grid = GridSearchCV(estimator=model, param_grid= parametreler, cv=cv, scoring="accuracy") #scoring kismina recall, f1 gibi degerler de verebiliriz
```

```python
grid
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
grid.fit(X_train, y_train)
```

```python
print(grid.best_params_,"\n") #en iyi parametreler neymis

print(grid.best_score_, "\n") #en iyi accuracy skoru neymis
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_pred, y_test)
print("score: ", score,"\n")

print(classification_report(y_test, y_pred))

print("confusion_matrix: \n",confusion_matrix(y_test, y_pred))
```

### Random Search CV

```python
from sklearn.model_selection import RandomizedSearchCV
```

### Grid Search CV, denemek istediğiniz her bir hiperparametre değerinin tüm olası kombinasyonlarını sistematik bir şekilde denerken, Random Search CV, belirlediğiniz aralıklardan rastgele kombinasyonlar seçerek dener.<br><br>__Hiperparametre Uzayını Tanımlama__: Optimizasyon yapmak istediğiniz hiperparametreler ve her biri için denemek istediğiniz değer aralıklarını veya dağılımlarını belirlersiniz.-->ornegin: C=[0.1-100] arasi<br><br>__Rastgele Kombinasyon Seçimi__: Belirlediğiniz aralıklardan, denemek istediğiniz sayıda rastgele kombinasyon seçersiniz. Örneğin, 100 kombinasyon denemeye karar verirseniz, algoritma bu aralıklardan 100 rastgele kombinasyon seçer.

### __Random Search CV__ ise, özellikle optimizasyon yapacağınız hiperparametre sayısının çok olduğu ve sınırlı bir zamanınız olduğu durumlarda daha verimlidir. Pratikte, Grid Search'e göre daha kısa sürede daha iyi veya benzer sonuçlar verdiği sıkça görülmüştür. Bu yüzden büyük projelerde genellikle Random Search tercih edilir.

```python
model_random= LogisticRegression()
```

```python
random_cv = RandomizedSearchCV(estimator= model_random, param_distributions=parametreler, cv=5, n_iter=10, scoring="accuracy")
#n_iter-->denedigi random sayisi ,,,,, 
```

```python
random_cv.fit(X_train, y_train)
```

```python
print(random_cv.best_params_, "\n", "score: ", random_cv.best_score_)
```

```python
y_pred2 = random_cv.predict(X_test)
```

```python
score = accuracy_score(y_pred2, y_test)
print("score: ", score,"\n")

print(classification_report(y_test, y_pred2))

print("confusion_matrix: \n",confusion_matrix(y_test, y_pred2))
```
